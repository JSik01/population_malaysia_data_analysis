```python
import pandas as pd
import matplotlib.pyplot as plt

import sys
from pathlib import Path

ROOT = Path().resolve().parents[0]
sys.path.append(str(ROOT))

from paths import DATA_FILE

population_data = pd.read_csv(DATA_FILE)
population_data["year"] = pd.to_datetime(population_data["date"]).dt.year

population_data = population_data[(population_data["age"] == "overall") &
                                          (population_data["ethnicity"] == "overall") &
                                          (population_data["sex"] == "both")]
```

```python
grouped = population_data.pivot_table("population", index = "state",
                                         columns = ["year"], aggfunc="sum")
grouped = grouped.T
grouped = grouped.fillna(0)
grouped["Malaysia"] = grouped.sum(axis=1)
grouped
```

```python
malaysia_data = grouped["Malaysia"]
malaysia_data
```

Kalman filter uses a series of measurements over time to produce unknown estimates using predicting and updating steps

Local Linear Trend (LLT) is the simplest. It decomposes a time series model into a level component, current value, and trend component, rate of change

First, fit a local linear model after converting year into index from start of year

```python
import statsmodels.api as sm

malaysia_data.index = pd.PeriodIndex(malaysia_data.index, freq='Y').to_timestamp(how='start')

model = sm.tsa.UnobservedComponents(malaysia_data, level= 'local linear trend')
result = model.fit()
print(result.summary())
```

Next, plot slope and trend components

```python
result.plot_components(figsize = (12, 12), legend_loc= 'upper left');
```

From trend plot, Malaysia's population growth increased from 1970 to around 2000. Then, the growth stabilsied to 2010 before dipping. The growth is higher but is below previous values after 2020.

Forecasting with Kalman filter

```python
forecast = result.get_forecast(steps=20)
pred = forecast.predicted_mean
ci = forecast.conf_int()

# Align to year-start
pred.index = pred.index.to_period('Y').to_timestamp(how='start')
ci.index = ci.index.to_period('Y').to_timestamp(how='start')

import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.plot(malaysia_data, label='Observed')
plt.plot(pred, label='Forecast')
plt.fill_between(ci.index, ci.iloc[:,0], ci.iloc[:,1], alpha=0.2)
plt.legend()
plt.show()
```

Unfortunately, due to timing of date of using start-of-year, there is a small gap between final observed date (2025-01-01) and first forecast date (2026-01-01)

The forecast line is linear with widening confidence band, reflecting non-seasonal population data with growing uncertainty

Test the model using plot_diagnostics()

```python
result.plot_diagnostics(figsize=(12, 12));
```

Standardized residuals over time and normal Q-Q plots show a structural break around 2020

Histogram plus estimated density of standardized residuals has a higher peak from standard normal distribution curve, but the residuals are not skewed too much

Correlogram plot shows there is no observable trend and all other lags after lag 1 is within confidence band. There is no serious autocorrelation.

Confirm that there is unlikely to have an autocorrelation problem using Ljung-Box test.

```python
from statsmodels.stats.diagnostic import acorr_ljungbox

acorr_ljungbox(result.resid, lags=[5, 10, 15], return_df=True)
```

As p-value is close to 1, there is not enough evidence to reject null hypothesis that there is no serial correlation in Malaysia's population data

Local linear trend state-space model is adequate to model Malaysia's population data

The Kalman filter state-space model forecast model is very likely to be valid based on diagnostic plots
