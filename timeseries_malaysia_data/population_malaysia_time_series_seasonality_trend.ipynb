```python
import pandas as pd
import matplotlib.pyplot as plt

import sys
from pathlib import Path

ROOT = Path().resolve().parents[0]
sys.path.append(str(ROOT))

from paths import DATA_FILE

population_data = pd.read_csv(DATA_FILE)
population_data["year"] = pd.to_datetime(population_data["date"]).dt.year

population_data = population_data[(population_data["age"] == "overall") &
                                          (population_data["ethnicity"] == "overall") &
                                          (population_data["sex"] == "both")]
```

```python
grouped = population_data.pivot_table("population", index = "state",
                                         columns = ["year"], aggfunc="sum")
grouped = grouped.T
grouped = grouped.fillna(0)
grouped["Malaysia"] = grouped.sum(axis=1)
grouped
```

```python
malaysia_data = grouped["Malaysia"]
malaysia_data
```

Seasonality is unlkely to be present in an annual population data, so we skip this part

The data is likely to be stochastic (random) instead of deterministic (can be expressed as an equation), but a check for deterministic won't hurt

```python
import statsmodels.api as sm
import numpy as np

y = malaysia_data.values
t = np.arange(len(malaysia_data))

X = sm.add_constant(t)
model = sm.OLS(y, X).fit()

print(model.summary())
```

Population data has strong upward linear deterministic trend, as p-value for x1 is very low

```python
from statsmodels.tsa.stattools import adfuller

adf_resid = adfuller(model.resid)
print("ADF statistics value:", adf_resid[0])
print("p-value:", adf_resid[1])
```

However, ADF test on residual reveals that residuals are non-stationary, implying the population data is dominated by stochastic trend.

Confirming non-stationarity around deterministic trend using ADF with regression = 'ct' (constant + linear term) on original data

```python
from statsmodels.tsa.stattools import adfuller

pvalue_adf = adfuller(malaysia_data, regression='ct')[1]
print(pvalue_adf)
```

p-value is 0.91, confirming that the trend for population data is very likely to be stochastic

From ZA test (I(2) behavior) and trend test (stochastic trend), we conclude that Kalman filter plot is the best model
