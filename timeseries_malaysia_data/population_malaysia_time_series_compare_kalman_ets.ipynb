Let's compare Kalman filter and ETS

A limitation is no external factors are considered

```python
import pandas as pd
import matplotlib.pyplot as plt

import sys
from pathlib import Path

ROOT = Path().resolve().parents[0]
sys.path.append(str(ROOT))

from paths import DATA_FILE

population_data = pd.read_csv(DATA_FILE)
population_data["year"] = pd.to_datetime(population_data["date"]).dt.year

population_data = population_data[(population_data["age"] == "overall") &
                                          (population_data["ethnicity"] == "overall") &
                                          (population_data["sex"] == "both")]
```

```python
grouped = population_data.pivot_table("population", index = "state",
                                         columns = ["year"], aggfunc="sum")
grouped = grouped.T
grouped = grouped.fillna(0)
grouped["Malaysia"] = grouped.sum(axis=1)
grouped
```

```python
malaysia_data = grouped["Malaysia"]
malaysia_data
```

```python
import statsmodels.api as sm

malaysia_data.index = pd.PeriodIndex(malaysia_data.index, freq='Y').to_timestamp(how='start')

model = sm.tsa.UnobservedComponents(malaysia_data, level= 'local linear trend')
result = model.fit()
```

```python
result.summary()
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing

ets = ExponentialSmoothing(malaysia_data, trend = "add", damped_trend = False).fit()
ets_forecast = ets.forecast(20)
```

```python
ets.summary()
```

A simpler ETS model naturally has lower AIC and BIC

Plot forecast of both models

```python
forecast = result.get_forecast(steps=20)
pred = forecast.predicted_mean
ci = forecast.conf_int()

# Align to year-start
pred.index = pred.index.to_period('Y').to_timestamp(how='start')
ci.index = ci.index.to_period('Y').to_timestamp(how='start')

import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.plot(malaysia_data, label='Observed')
plt.plot(pred, label='Forecast')
plt.fill_between(ci.index, ci.iloc[:,0], ci.iloc[:,1], alpha=0.2, color="pink")

sigma = ets.resid.std()

# 95% CI
lower = ets_forecast - 1.96 * sigma
upper = ets_forecast + 1.96 * sigma

plt.plot(ets_forecast, label='ETS Forecast')

plt.fill_between(
    ets_forecast.index,
    lower,
    upper,
    color='gray',
    alpha=0.3,
    label='95% CI'
)

# Zoom in
plt.ylim(33000, 41000)
# Number of days from 1 January 1970
plt.xlim(18000, 28000)
plt.legend()
plt.show()
```

A deterministic trend and failure to detect structural breaks leads to ETS having higher population forecast than Kalman filter even when AIC and BIC of ETS is lower than Kalman filter
